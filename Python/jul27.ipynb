```python
day=int(input("Enter the Day number : "))
match day:
    case 1:print("Monday")
    case 2:print("Tuesday")
    case 3:print("Wednesday")
    case 4:print("Thursday")
    case 5:print("Friday")
    case 6:print("Saturday")
    case _:print("Invalid Inputs")
```

```python
a,b=map(int,input("Enter the 2 digits : ").split())
c=input("Enter the Operation (*,/,-,+): ")
match c:
    case '+' :print(a+b)
    case '-' :print(a-b)
    case '/':print(a/b)
    case '*' :print(a*b)
    case _:print("Null Operation")
```
