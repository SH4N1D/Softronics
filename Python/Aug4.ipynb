```python
a=list(map(int,input('Enter the Marks : ').split()))
i=0
while i < len(a):
    if a[i]<35:
        print(f'{a[i]} : fail')
    else:
        print(f'{a[i]} : Pass')
    i+=1
b=sum(a)
avg=b/len(a)
if avg>90:
    print('Grade A')
elif avg >75:
    print('Grade B')
else:
    print('Grade D')
    
```
