```python
def listcreate():
    a=list(map(int,input("Enter the List Elements : ").split()))
    return a
```

```python
def duplicateremover(x):
    z=set(x)
    return list(z)
```

```python
a=listcreate()
print(duplicateremover(a))
```

```python
def wordcount(s):
    return len(list(s.split()))
```

```python
t=input("Enter the String : ")
print(wordcount(t))
```

```python
def palindrome(s):
    if s==s[::-1]:
        return "The String is a Palindrome"
    else:
        return "The String is Not a Palindrome"
```

```python
palindrome('malayalam')
```

```python
def listofnumbers():
    d=list(map(int,input("Enter the List Elements : ").split()))
    return sum(d)
```

```python
listofnumbers()
```

```python
def avg():
    a,b,c=map(int,input("Enter the 3 Elements : ").split())
    u=a+b+c/3
    return u
```
